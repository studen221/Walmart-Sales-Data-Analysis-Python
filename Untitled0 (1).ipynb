```python
# 1. What is the total sales volume across all walmart stores ?

df ['quantity_sold'].sum()
print("Total number of quantity_sold:-",df ['quantity_sold'].sum())
```

```python
# 2. What is the average unit price of all products sold?

from numpy import average
average_price= df.groupby('product_name')['unit_price'].mean()
#plot bar graph

import matplotlib.pyplot as plt

# Plot Bar Graph
plt.figure(figsize=(10,6))
bars = average_price.plot(kind='bar', color='skyblue')

plt.title('Average Price per Product')
plt.xlabel('Product Name')
plt.ylabel('Average Unit Price')
plt.xticks(rotation=45, ha='right')

# Add Data Values on Bars
for index, value in enumerate(average_price):
    plt.text(index, value, f'{value:.2f}',
             ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

```python
# 3.Which stores are underperforming in terms of sales volume?

sales_by_store = df.groupby('store_id')['quantity_sold'].sum().sort_values()

print(sales_by_store)
```

```python
threshold= sales_by_store.median()
underperforming_stores = sales_by_store[sales_by_store < threshold]
print( "underperforming stores:")
print (underperforming_stores)
```

```python
# 4. What count of transactions involved promotions?

import seaborn as sns
import matplotlib.pyplot as plt
ax=sns.countplot(x='promotion_applied', data=df)
for bar in ax.containers:
  ax.bar_label(bar)
```

```python
#5. What is the total revenue generate from sales during promotional events?

# Calculate revenue
df['revenue'] = df['quantity_sold'] * df['unit_price']

# Calculate total revenue where promotion was applied
promo_revenue = df[df['promotion_applied'] == True]['revenue'].sum()

print(promo_revenue)
```

```python
# 6. Which product categories see the largest increase in demand during holidays ?
import matplotlib.pyplot as plt

# Filter data for holidays
holiday_sales = df[df['holiday_indicator'] == True]

# Group by product name and calculate total quantity sold
holiday_demand = holiday_sales.groupby('product_name')['quantity_sold'].sum()

# Display results
print(holiday_demand)

# Plot the bar graph
plt.figure(figsize=(12,6))
ax = holiday_demand.plot(kind='bar', color='skyblue', edgecolor='black')

# Add data labels
for index, value in enumerate(holiday_demand):
    plt.text(index, value, str(value),
             ha='center', va='bottom', fontsize=10)

# Customize the plot
plt.title('Total Quantity Sold During Holidays by Product', fontsize=14)
plt.xlabel('Product Name', fontsize=12)
plt.ylabel('Total Quantity Sold', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
# 7. What is the correlation between weather conditions and sales performance?

import matplotlib.pyplot as plt

# Group by weather conditions and calculate total revenue
a = df.groupby('weather_conditions')['revenue'].sum()

print(a)

# Plot
plt.figure(figsize=(10,6))
a.plot(kind='bar', color='skyblue', edgecolor='black')

# Data labels
for i, value in enumerate(a):
    plt.text(i, value, f'{value:.2f}',
             ha='center', va='bottom')

plt.title("Revenue by Weather Conditions")
plt.xlabel("Weather Conditions")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
# 8. how does customer loyalty level affect purchasing patterns

loyalty_analysis = df.groupby('customer_loyalty_level').agg({
    'revenue': ['sum', 'mean'],
    'customer_id': 'nunique',
    'transaction_id': 'count'
})

print(loyalty_analysis)
```

```python
# perform groupby and aggregation

loyalty_analysis = df.groupby('customer_loyalty_level').agg({
    'revenue': ['sum', 'mean'],
    'customer_id': 'nunique',
    'transaction_id': 'count'
}).reset_index()

loyalty_analysis_sorted = loyalty_analysis.sort_values(
    by=('revenue', 'sum'),
    ascending=False
)

print(loyalty_analysis_sorted)
```

```python
 # 9. calculate the monthly sales and find out the trends

df.info()
```

```python
df['transaction_date'] = pd.to_datetime(df['transaction_date'])
```

```python
df['Month'] = df['transaction_date'].dt.month_name()
```

```python
df['Month_Number'] = df['transaction_date'].dt.month
```

```python
df['Month'].unique()
```

```python
#step 2 group by month and calculate total revenue
monthly_revenue = df.groupby('Month')['revenue'].sum().reset_index()

#step 3 plot the trend line

import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.plot(monthly_revenue['Month'],
         monthly_revenue['revenue'],
         marker='o',
         linestyle='--',
         color='b')

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.grid()
plt.show()

# Monthly Sales
monthly_sales_Walmart = df.groupby('Month')['revenue'].sum().sort_values()

monthly_sales_Walmart
```

```python
#(10) which product has the best peformance in tems of volume sold

best_product = df.groupby('product_name')['quantity_sold'].sum()
best_product
```

```python
#plot the bar graph
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

best_product.plot(
    kind='bar',
    color='purple',
    edgecolor='black'
)

# Customize the plot
plt.title('Best Selling Products', fontsize=14)
plt.xlabel('Product Name', fontsize=12)
plt.ylabel('Quantity Sold', fontsize=12)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

# Add data labels
for index, value in enumerate(best_product):
    plt.text(
        index,
        value + 0.01 * max(best_product),
        f'{value:.0f}',
        ha='center',
        va='bottom',
        fontsize=10,
        color='black'
    )

# Show the plot
plt.show()
```
